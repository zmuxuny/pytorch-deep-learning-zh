好，现在来看 **vector（向量）**。

向量是一个一维张量，但可以包含多个数字。

例如，你可以用向量 `[3, 2]` 表示房子的 `[bedrooms, bathrooms]`（卧室数、卫生间数）。也可以用 `[3, 2, 2]` 表示 `[bedrooms, bathrooms, car_parks]`（卧室数、卫生间数、车位数）。

这里的重要趋势是：向量在表示内容上非常灵活（张量同样如此）。

````python
# 向量（Vector）
vector = torch.tensor([7, 7])
vector
````

很好，`vector` 现在包含两个 7，这是我最喜欢的数字。

你觉得它会有多少个维度？

````python
# 检查 vector 的维度数
vector.ndim
````

嗯，有点奇怪，`vector` 里有两个数字，但只有一个维度。

我告诉你一个小技巧。

在 PyTorch 里，你可以通过最外层方括号（`[`）的层数判断张量有多少维度，只需要数一侧即可。

`vector` 有几层方括号？

````python
# 检查维度数
MATRIX.ndim
````

`MATRIX` 有两个维度（你有数最外层一侧方括号的层数吗？）

我们得到输出 `torch.Size([2, 2])`，因为 `MATRIX` 的深度为 2、宽度也为 2。

### 随机张量

我们已经知道张量用于表示某种形式的数据。

而机器学习模型（如神经网络）会对张量进行操作并在其中寻找模式。

但在用 PyTorch 构建机器学习模型时，你很少会手工创建张量（就像我们目前这样）。

相反，模型通常从大规模随机张量开始，然后在处理数据的过程中不断调整这些随机数，使其更好地表示数据。

本质上是：

`从随机数开始 -> 观察数据 -> 更新随机数 -> 再观察数据 -> 再更新随机数...`

作为数据科学家，你可以定义模型如何开始（初始化）、如何看数据（表示）以及如何更新（优化）这些随机数。

我们后面会动手实践这些步骤。

现在先看看如何创建一个随机数张量。

我们可以使用 [`torch.rand()`](https://pytorch.org/docs/stable/generated/torch.rand.html) 并传入 `size` 参数。

````python
# 创建一个大小为 (3, 4) 的随机张量
random_tensor = torch.rand(size=(3, 4))
random_tensor, random_tensor.dtype
````

`torch.rand()` 的灵活之处在于我们可以按需调整 `size`。

例如，如果你想创建一个常见图像 shape 的随机张量 `[224, 224, 3]`（`[height, width, color_channels]`）。

````python
# 创建一个大小为 (224, 224, 3) 的随机张量
random_image_size_tensor = torch.rand(size=(224, 224, 3))
random_image_size_tensor.shape, random_image_size_tensor.ndim
````

### 全 0 与全 1

有时候你会希望用 0 或 1 来填充张量。

这在 mask（掩码）场景中非常常见（例如把某些位置置 0，告诉模型这些值不需要学习）。

让我们用 [`torch.zeros()`](https://pytorch.org/docs/stable/generated/torch.zeros.html) 创建一个全 0 张量。

同样，`size` 参数依然关键。

````python
# 创建一个全 0 张量
zeros = torch.zeros(size=(3, 4))
zeros, zeros.dtype
````

同理，我们也可以创建全 1 张量，只需要使用 [`torch.ones()`](https://pytorch.org/docs/stable/generated/torch.ones.html)。

````python
# 创建一个全 1 张量
ones = torch.ones(size=(3, 4))
ones, ones.dtype
````

### 创建数值区间与同形张量

有时候你可能想要一个数值区间，例如 1 到 10，或 0 到 100。

你可以使用 `torch.arange(start, end, step)`。

其中：
* `start` = 区间起点（例如 0）
* `end` = 区间终点（例如 10）
* `step` = 步长，即每个值之间间隔多少（例如 1）

> **注意：** 在 Python 中，你可以使用 `range()` 创建区间；但在 PyTorch 中，`torch.range()` 已被弃用，未来可能报错。

````python
# 使用 torch.arange()，torch.range() 已弃用
zero_to_ten_deprecated = torch.range(0, 10) # 注意：未来可能报错

# 创建 0 到 10 的数值区间
zero_to_ten = torch.arange(start=0, end=10, step=1)
zero_to_ten
````

有时你会希望创建一个张量，它与另一个张量 shape 相同，但内容类型不同。

例如，创建一个与前一个张量 shape 相同的全 0 张量。

这时可以使用 [`torch.zeros_like(input)`](https://pytorch.org/docs/stable/generated/torch.zeros_like.html) 或 [`torch.ones_like(input)`](https://pytorch.org/docs/1.9.1/generated/torch.ones_like.html)。它们会分别返回与 `input` 同 shape 的全 0 或全 1 张量。

````python
# 也可以创建与另一个张量同形状的全 0 张量
ten_zeros = torch.zeros_like(input=zero_to_ten) # shape 将保持一致
ten_zeros
````

### 张量数据类型（datatypes）

PyTorch 提供了许多不同的[张量数据类型](https://pytorch.org/docs/stable/tensors.html#data-types)。

有些类型更适合 CPU，有些更适合 GPU。

理解每种类型的使用场景需要一些时间。

通常如果你看到 `torch.cuda`，说明这个张量在 GPU 上使用（因为 Nvidia GPU 使用名为 CUDA 的计算工具链）。

最常见的类型（通常也是默认类型）是 `torch.float32` 或 `torch.float`。

它被称为“32-bit floating point（32 位浮点数）”。

此外还有 16 位浮点（`torch.float16` 或 `torch.half`）和 64 位浮点（`torch.float64` 或 `torch.double`）。

更容易混淆的是，还有 8 位、16 位、32 位和 64 位整数类型。

以及更多类型！

````python
# 张量的默认数据类型是 float32
float_32_tensor = torch.tensor([3.0, 6.0, 9.0],
                               dtype=None, # 默认为 None，即 torch.float32 或传入的数据类型
                               device=None, # 默认为 None，即使用默认张量设备类型
                               requires_grad=False) # 若为 True，会记录在张量上执行的运算

float_32_tensor.shape, float_32_tensor.dtype, float_32_tensor.device
````

除了 shape 问题（张量形状不匹配）之外，在 PyTorch 中你最常遇到的另外两类问题是数据类型和设备问题。

例如，一个张量是 `torch.float32`，另一个是 `torch.float16`（PyTorch 通常更希望参与计算的张量格式一致）。

或者一个张量在 CPU 上，另一个在 GPU 上（PyTorch 更希望参与计算的张量位于同一设备）。

后面我们还会继续讲设备相关内容。

现在先创建一个 `dtype=torch.float16` 的张量。

````python
float_16_tensor = torch.tensor([3.0, 6.0, 9.0],
                               dtype=torch.float16) # torch.half 也可以

float_16_tensor.dtype
````

## 从张量中获取信息

当你创建了张量（或由他人、某个 PyTorch 模块为你创建了张量）后，通常会想了解它的一些信息。

我们之前已经见过这些属性，其中最常用的三个是：
* `shape` - 张量的形状是什么？（某些操作要求特定形状规则）
* `dtype` - 张量内部元素以什么数据类型存储？
* `device` - 张量存储在哪个设备上？（通常是 GPU 或 CPU）

我们创建一个随机张量并查看其细节。

````python
# 创建一个张量
some_tensor = torch.rand(3, 4)

# 查看它的详细信息
print(some_tensor)
print(f"张量形状: {some_tensor.shape}")
print(f"张量数据类型: {some_tensor.dtype}")
print(f"张量存储设备: {some_tensor.device}") # 默认会是 CPU
````

## 操作张量（tensor operations）

在深度学习中，数据（图像、文本、视频、音频、蛋白质结构等）都会被表示为张量。

模型通过分析这些张量，并在其上执行一系列操作（可能是上百万次）来学习输入数据中的模式表示。

这些操作通常是以下几类的组合：
* 加法
* 减法
* 乘法（逐元素）
* 除法
* 矩阵乘法

基本就这些。当然还有一些其他操作，但这些是神经网络最基础的积木。

只要把这些积木以正确方式组合起来，你就能构建非常复杂的神经网络（就像乐高一样）。

### 基础操作

先从几个最基础的操作开始：加法（`+`）、减法（`-`）、乘法（`*`）。

它们的行为和你直觉中的一致。

````python
# 创建一个张量并给它加一个数
tensor = torch.tensor([1, 2, 3])
tensor + 10
````

````python
# 乘以 10
tensor * 10
````

注意，上面的张量值并没有变成 `tensor([110, 120, 130])`，因为张量中的值只有在重新赋值后才会改变。

````python
# 张量不会变化，除非重新赋值
tensor
````

我们来减一个数，这次把结果重新赋值给 `tensor` 变量。

````python
# 减法并重新赋值
tensor = tensor - 10
tensor
````

````python
# 加法并重新赋值
tensor = tensor + 10
tensor
````

### 矩阵乘法（非常关键）

在机器学习和深度学习算法（如神经网络）中，最常见的操作之一是[矩阵乘法](https://www.mathsisfun.com/algebra/matrix-multiplying.html)。

PyTorch 在 [`torch.matmul()`](https://pytorch.org/docs/stable/generated/torch.matmul.html) 方法中实现了矩阵乘法功能。

你需要记住矩阵乘法的两条核心规则：

1. **内维度** 必须匹配：
  * `(3, 2) @ (3, 2)` 不能工作
  * `(2, 3) @ (3, 2)` 可以工作
  * `(3, 2) @ (2, 3)` 可以工作
2. 结果矩阵的 shape 由 **外维度** 决定：
 * `(2, 3) @ (3, 2)` -> `(2, 2)`
 * `(3, 2) @ (2, 3)` -> `(3, 3)`

> **注意：** 在 Python 中，`@` 是矩阵乘法符号。

> **资源：** 你可以在 [PyTorch 文档](https://pytorch.org/docs/stable/generated/torch.matmul.html) 中查看使用 `torch.matmul()` 的完整规则。

我们创建一个张量，并对其进行逐元素乘法和矩阵乘法。


````python
import torch
tensor = torch.tensor([1, 2, 3])
tensor.shape
````

逐元素乘法和矩阵乘法的关键区别在于是否进行求和。

对于值为 `[1, 2, 3]` 的 `tensor` 变量：

| 运算 | 计算方式 | 代码 |
| ----- | ----- | ----- |
| **逐元素乘法** | `[1*1, 2*2, 3*3]` = `[1, 4, 9]` | `tensor * tensor` |
| **矩阵乘法** | `[1*1 + 2*2 + 3*3]` = `[14]` | `tensor.matmul(tensor)` |


````python
# 逐元素矩阵乘法
tensor * tensor
````

````python
# 矩阵乘法
torch.matmul(tensor, tensor)
````

````python
# 也可以使用 "@" 符号做矩阵乘法，但不推荐
tensor @ tensor
````

你可以手写矩阵乘法，但不推荐。

内置的 `torch.matmul()` 更快。

````python
%%time
# 手动实现矩阵乘法
# （尽量避免使用 for 循环做此类运算，它们计算开销较高）
value = 0
for i in range(len(tensor)):
  value += tensor[i] * tensor[i]
value
````

````python
%%time
torch.matmul(tensor, tensor)
````

## 深度学习中最常见的错误之一（形状错误）

由于深度学习的大量计算都依赖矩阵运算，而矩阵对可组合的 shape 和尺寸有严格要求，因此 shape 不匹配是深度学习中最常见的错误之一。

````python
# shape 必须匹配正确
tensor_A = torch.tensor([[1, 2],
                         [3, 4],
                         [5, 6]], dtype=torch.float32)

tensor_B = torch.tensor([[7, 10],
                         [8, 11], 
                         [9, 12]], dtype=torch.float32)

torch.matmul(tensor_A, tensor_B) # （这里会报错）
````

要让 `tensor_A` 和 `tensor_B` 可以做矩阵乘法，需要让它们的内维度一致。

一种方法是使用 **transpose（转置）**（交换给定张量的维度）。

在 PyTorch 中可以这样做：
* `torch.transpose(input, dim0, dim1)` - 其中 `input` 是要转置的张量，`dim0` 和 `dim1` 是要交换的维度。
* `tensor.T` - 其中 `tensor` 是要转置的张量。

我们先试第二种。

````python
# 查看 tensor_A 和 tensor_B
print(tensor_A)
print(tensor_B)
````

````python
# 查看 tensor_A 和 tensor_B.T
print(tensor_A)
print(tensor_B.T)
````

````python
# 当 tensor_B 被转置后，运算就能正常进行
print(f"原始形状: tensor_A = {tensor_A.shape}, tensor_B = {tensor_B.shape}\n")
print(f"新形状: tensor_A = {tensor_A.shape}（与上方相同）, tensor_B.T = {tensor_B.T.shape}\n")
print(f"执行乘法: {tensor_A.shape} * {tensor_B.T.shape} <- 内维度匹配\n")
print("输出:\n")
output = torch.matmul(tensor_A, tensor_B.T)
print(output) 
print(f"\n输出形状: {output.shape}")
````

你也可以使用 [`torch.mm()`](https://pytorch.org/docs/stable/generated/torch.mm.html)，它是 `torch.matmul()` 的简写。

````python
# torch.mm 是 matmul 的快捷写法
torch.mm(tensor_A, tensor_B.T)
````

如果不做转置，矩阵乘法规则就不满足，因此会像上面那样报错。

来看一个可视化示例：

![矩阵乘法可视化示意](https://github.com/mrdbourke/pytorch-deep-learning/raw/main/images/00-matrix-multiply-crop.gif)

你也可以在 http://matrixmultiplication.xyz/ 创建这样的矩阵乘法可视化。

> **注意：** 这种矩阵乘法也被称为两个矩阵的 [**dot product（点积）**](https://www.mathsisfun.com/algebra/vectors-dot-product.html)。


神经网络中充满了矩阵乘法和点积。

[`torch.nn.Linear()`](https://pytorch.org/docs/1.9.1/generated/torch.nn.Linear.html) 模块（后面会看到它的实际用法），也叫前馈层（feed-forward layer）或全连接层（fully connected layer），实现了输入 `x` 与权重矩阵 `A` 之间的矩阵乘法。

$$
y = x\cdot{A^T} + b
$$

其中：
* `x` 是层的输入（深度学习通常是多层堆叠而成，比如 `torch.nn.Linear()` 与其他层）。
* `A` 是该层创建的权重矩阵，初始通常是随机数，并会在神经网络学习过程中不断调整，以更好表示数据模式（注意这里的 `T`，表示权重矩阵会被转置）。

````python
# 线性层从随机权重矩阵开始，因此我们先设置可复现性（后面会详细讲）
torch.manual_seed(42)
# 这里内部会用到矩阵乘法
linear = torch.nn.Linear(in_features=2, # in_features = 与输入内维度匹配
                         out_features=6) # out_features = 描述外维度输出
x = tensor_A
output = linear(x)
print(f"输入形状: {x.shape}\n")
print(f"输出:\n{output}\n\n输出形状: {output.shape}")
````

> **问题：** 如果把上面的 `in_features` 从 2 改成 3，会发生什么？会报错吗？你该如何调整输入 `x` 的 shape 来适配这个错误？提示：上面我们对 `tensor_B` 做了什么操作？

但当你多练几次，甚至拆开看过几个神经网络之后，你会发现它几乎无处不在。

记住，矩阵乘法非常关键。

### Reshaping、stacking、squeezing 与 unsqueezing

很多时候你希望改变张量的形状或维度，但不改变其中的数值。

常见方法有：

| 方法 | 一句话说明 |
| ----- | ----- |
| [`torch.reshape(input, shape)`](https://pytorch.org/docs/stable/generated/torch.reshape.html#torch.reshape) | 将 `input` 重塑为 `shape`（若兼容）；也可用 `torch.Tensor.reshape()`。 |
| [`Tensor.view(shape)`](https://pytorch.org/docs/stable/generated/torch.Tensor.view.html) | 返回原始张量在新 `shape` 下的视图，和原张量共享同一份数据。 |
| [`torch.stack(tensors, dim=0)`](https://pytorch.org/docs/1.9.1/generated/torch.stack.html) | 沿新维度 `dim` 拼接一组 `tensors`，所有张量必须同尺寸。 |
| [`torch.squeeze(input)`](https://pytorch.org/docs/stable/generated/torch.squeeze.html) | 压缩 `input`，移除值为 `1` 的所有维度。 |
| [`torch.unsqueeze(input, dim)`](https://pytorch.org/docs/1.9.1/generated/torch.unsqueeze.html) | 在 `dim` 位置新增一个值为 `1` 的维度并返回。 | 
| [`torch.permute(input, dims)`](https://pytorch.org/docs/stable/generated/torch.permute.html) | 返回原始 `input` 的一个 *view*，其维度顺序按 `dims` 重排。 | 

为什么要做这些？

因为深度学习模型（神经网络）本质上就是不断操作张量。而矩阵乘法有严格 shape 规则，一旦 shape 不匹配就会报错。这些方法能帮助你确保张量中的正确元素与其他张量中的正确元素进行运算。

我们来试试。

先创建一个张量。

````python
# 创建一个张量
import torch
x = torch.arange(1., 8.)
x, x.shape
````

现在用 `torch.reshape()` 增加一个额外维度。

````python
# 增加一个额外维度
x_reshaped = x.reshape(1, 7)
x_reshaped, x_reshaped.shape
````

我们也可以用 `torch.view()` 改变视图。

````python
# 改变视图（保留原始数据，但改变查看方式）
# 更多说明: https://stackoverflow.com/a/54507446/7900723
z = x.view(1, 7)
z, z.shape
````

但要记住，用 `torch.view()` 改变张量视图，本质上只是为 *同一个* 张量创建了新视图。

因此改变这个视图，也会改变原始张量。

````python
# 修改 z 会同步影响 x
z[:, 0] = 5
z, x
````

如果我们想把新张量自身堆叠多次，可以使用 `torch.stack()`。

````python
# 将张量在新维度上堆叠
x_stacked = torch.stack([x, x, x, x], dim=0) # 可尝试把 dim 改为 1 观察变化
x_stacked
````

那如果要移除张量中所有单维度呢？

可以使用 `torch.squeeze()`（我把它记成把张量“挤压”到只保留大于 1 的维度）。

````python
print(f"原始张量: {x_reshaped}")
print(f"原始形状: {x_reshaped.shape}")

# 从 x_reshaped 中移除额外维度
x_squeezed = x_reshaped.squeeze()
print(f"\n新张量: {x_squeezed}")
print(f"新形状: {x_squeezed.shape}")
````

反过来，如果要做 `torch.squeeze()` 的逆操作，可以用 `torch.unsqueeze()` 在指定索引处增加一个值为 1 的维度。

````python
print(f"原始张量: {x_squeezed}")
print(f"原始形状: {x_squeezed.shape}")

## 使用 unsqueeze 增加额外维度
x_unsqueezed = x_squeezed.unsqueeze(dim=0)
print(f"\n新张量: {x_unsqueezed}")
print(f"新形状: {x_unsqueezed.shape}")
````

你还可以用 `torch.permute(input, dims)` 重排轴顺序，其中 `input` 会被转换为一个具有新 `dims` 的 *view*。

````python
# 创建特定形状的张量
x_original = torch.rand(size=(224, 224, 3))

# 对原始张量做 permute 以重排轴顺序
x_permuted = x_original.permute(2, 0, 1) # 轴变换: 0->1, 1->2, 2->0

print(f"原始形状: {x_original.shape}")
print(f"新形状: {x_permuted.shape}")
````

> **注意**: 因为 permute 返回的是一个 *view*（与原始张量共享数据），所以重排后张量的值与原始张量相同；若你修改该 view 的值，也会影响原始张量。

## 可复现性（尽量减少随机中的随机）

随着你对神经网络和机器学习理解加深，你会发现随机性在其中扮演了很大角色。

更准确地说，是伪随机。因为从设计上看，计算机本质上是确定性的（每一步都可预测），所以它生成的是“模拟随机性”（这个话题也有很多讨论，感兴趣你可以继续深入）。

这与神经网络和深度学习有什么关系？

我们前面提到过，神经网络通常从随机数出发去描述数据模式（这些初始描述通常很粗糙），再通过张量运算（以及一些我们还没讲到的机制）不断改进这些随机数，以更好地表示数据中的模式。

简要来说：

``从随机数开始 -> 进行张量运算 -> 持续改进（反复迭代）``

````python
import torch

# 创建两个随机张量
random_tensor_A = torch.rand(3, 4)
random_tensor_B = torch.rand(3, 4)

print(f"Tensor A:\n{random_tensor_A}\n")
print(f"Tensor B:\n{random_tensor_B}\n")
print(f"Tensor A 与 Tensor B 是否相等？（任意位置）")
random_tensor_A == random_tensor_B
````

正如你预期的那样，这两个张量的值不同。

但如果你想创建两个值 *相同* 的随机张量呢？

也就是说它们仍然是随机值，但“随机风味”一致。

这时就要用到 [`torch.manual_seed(seed)`](https://pytorch.org/docs/stable/generated/torch.manual_seed.html)。其中 `seed` 是整数（比如 `42`，也可以是任意整数），用于固定随机性的生成轨迹。

我们来试试，创建一些带固定“风味”的随机张量。

````python
import torch
import random

# # 设置随机种子
RANDOM_SEED=42 # 可尝试改成其他值，观察下面数值变化
torch.manual_seed(seed=RANDOM_SEED) 
random_tensor_C = torch.rand(3, 4)

# 每次新调用 rand() 前都要重新设置种子
# 否则 tensor_D 会与 tensor_C 不同
torch.random.manual_seed(seed=RANDOM_SEED) # 可尝试注释掉此行观察变化
random_tensor_D = torch.rand(3, 4)

print(f"Tensor C:\n{random_tensor_C}\n")
print(f"Tensor D:\n{random_tensor_D}\n")
print(f"Tensor C 与 Tensor D 是否相等？（任意位置）")
random_tensor_C == random_tensor_D
````

很好！

看起来设置随机种子生效了。

### 2. 让 PyTorch 在 GPU 上运行

当你拿到可用 GPU 后，下一步是让 PyTorch 用它来存储数据（张量）和计算（张量运算）。

可以使用 [`torch.cuda`](https://pytorch.org/docs/stable/cuda.html) 包。

与其讲概念，不如直接试。

你可以用 [`torch.cuda.is_available()`](https://pytorch.org/docs/stable/generated/torch.cuda.is_available.html#torch.cuda.is_available) 检查 PyTorch 是否可访问 GPU。


````python
# 检查 GPU
import torch
torch.cuda.is_available()
````

如果上面输出 `True`，说明 PyTorch 能看到并使用 GPU；如果输出 `False`，说明当前看不到 GPU，需要回到安装步骤排查。

现在假设你希望代码能在 CPU *或* GPU 上运行（若 GPU 可用则自动使用）。

这样无论你自己还是别人运行代码，都能适配其计算设备。

我们创建一个 `device` 变量来保存可用设备类型。

````python
# 设置设备类型
device = "cuda" if torch.cuda.is_available() else "cpu"
device
````

如果上面输出 `"cuda"`，表示我们可以让全部 PyTorch 代码使用可用的 CUDA 设备（GPU）；如果输出 `"cpu"`，则代码会继续在 CPU 上运行。

> **注意：** 在 PyTorch 中，最佳实践是写 [**device agnostic code**](https://pytorch.org/docs/master/notes/cuda.html#device-agnostic-code)（设备无关代码），即代码既能在 CPU（总可用）运行，也能在 GPU（可用时）运行。

如果你想更快计算，可以用 GPU；如果想 *更快更多*，可以使用多 GPU。

你可以用 [`torch.cuda.device_count()`](https://pytorch.org/docs/stable/generated/torch.cuda.device_count.html#torch.cuda.device_count) 统计 PyTorch 可访问的 GPU 数量。

````python
# 统计设备数量
torch.cuda.device_count()
````

了解 PyTorch 可访问 GPU 的数量很有帮助，比如你希望某个任务跑在一个 GPU、另一个任务跑在另一个 GPU（PyTorch 也支持跨 *全部* GPU 运行一个任务）。

### 2.1 在 Apple Silicon 上运行 PyTorch

要在 Apple 的 M1/M2/M3 GPU 上运行 PyTorch，可以使用 [`torch.backends.mps`](https://pytorch.org/docs/stable/notes/mps.html) 模块。

请确保 macOS 和 PyTorch 版本都已更新。

你可以用 `torch.backends.mps.is_available()` 检查 PyTorch 是否可访问 GPU。

````python
# 检查 Apple Silicon GPU
import torch
torch.backends.mps.is_available() # 注意：如果你不是在 Mac 上运行，这里会是 False
````

````python
# 设置设备类型
device = "mps" if torch.backends.mps.is_available() else "cpu"
device
````

和前面一样，如果上面输出 `"mps"`，说明我们可以让 PyTorch 代码使用可用的 Apple Silicon GPU。

````python
if torch.cuda.is_available():
    device = "cuda" # 使用 NVIDIA GPU（若可用）
elif torch.backends.mps.is_available():
    device = "mps" # 使用 Apple Silicon GPU（若可用）
else:
    device = "cpu" # 没有 GPU 时默认使用 CPU
````

### 3. 把张量（和模型）放到 GPU 上

你可以对张量（模型后面会讲）调用 [`to(device)`](https://pytorch.org/docs/stable/generated/torch.Tensor.to.html) 把它放到指定设备上。这里的 `device` 是目标设备。

为什么要这样做？

GPU 的数值计算通常比 CPU 快很多；如果没有 GPU，由于我们写的是 **设备无关代码**（见上文），代码会自动回退到 CPU。

> **注意：** 用 `to(device)` 把张量放到 GPU（例如 `some_tensor.to(device)`）会返回该张量的一个副本，也就是 CPU 与 GPU 上都会有。若要覆盖原变量，需要重新赋值：
>
> `some_tensor = some_tensor.to(device)`

我们来创建一个张量并把它放到 GPU（如果可用）。

````python
# 创建张量（默认在 CPU）
tensor = torch.tensor([1, 2, 3])

# 张量当前不在 GPU
print(tensor, tensor.device)

# 将张量移动到 GPU（如果可用）
tensor_on_gpu = tensor.to(device)
tensor_on_gpu
````

如果你有可用 GPU，上面代码会输出类似：

```
tensor([1, 2, 3]) cpu
tensor([1, 2, 3], device='cuda:0')
```

注意第二个张量有 `device='cuda:0'`，表示它存放在第 0 张可用 GPU 上（GPU 从 0 开始编号；若有两张，则分别是 `'cuda:0'` 和 `'cuda:1'`，依此类推到 `'cuda:n'`）。


### 4. 把张量移回 CPU

如果我们想把张量移回 CPU 呢？

例如，当你想让张量与 NumPy 交互时就需要这样做（NumPy 不使用 GPU）。

先试试在 `tensor_on_gpu` 上直接调用 [`torch.Tensor.numpy()`](https://pytorch.org/docs/stable/generated/torch.Tensor.numpy.html)。

````python
# 如果张量在 GPU 上，无法直接转为 NumPy（这里会报错）
tensor_on_gpu.numpy()
````

正确做法是先用 [`Tensor.cpu()`](https://pytorch.org/docs/stable/generated/torch.Tensor.cpu.html) 把张量放回 CPU，再与 NumPy 配合。

这会把张量复制到 CPU 内存，使其可被 CPU/NumPy 使用。

````python
# 将张量复制回 cpu
tensor_back_on_cpu = tensor_on_gpu.cpu().numpy()
tensor_back_on_cpu
````

上面的结果是在 CPU 内存中返回了 GPU 张量的副本，因此原始张量仍然在 GPU 上。
