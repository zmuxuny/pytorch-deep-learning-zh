## 设备无关代码（在 CPU、GPU 或 MPS 上使用 PyTorch）

深度学习的大量工作都涉及对张量进行计算。

对张量的计算通常在 GPU（图形处理器，通常来自 NVIDIA）上比在 CPU（中央处理器）上快得多。

MPS 指的是 "Metal Performance Shader"，这是苹果的 GPU（M1、M1 Pro、M2 等）。

建议使用你手头最快的硬件进行训练，通常优先级是：NVIDIA GPU（`"cuda"`）> MPS 设备（`"mps"`）> CPU（`"cpu"`）。

* 关于如何让 [PyTorch 在 NVIDIA GPU 上运行（使用 CUDA）](https://pytorch.org/docs/stable/cuda.html) 的更多信息，请查看 [00. PyTorch Fundamentals 第 2 节：让 PyTorch 在 GPU 上运行](https://www.learnpytorch.io/00_pytorch_fundamentals/#2-getting-pytorch-to-run-on-the-gpu)。
* 关于使用 MPS 后端运行 PyTorch（在 Mac GPU 上运行 PyTorch）的更多信息，请查看 [PyTorch 文档](https://pytorch.org/docs/stable/notes/mps.html)。

> **注意：** 建议在工作流一开始就设置设备无关代码。

```python
# 设置设备无关代码
if torch.cuda.is_available():
    device = "cuda" # NVIDIA GPU
elif torch.backends.mps.is_available():
    device = "mps" # Apple GPU
else:
    device = "cpu" # 如果没有 NVIDIA GPU/Apple GPU，则默认使用 CPU

print(f"Using device: {device}")
```

## 神经网络

PyTorch 提供了非常全面的预置神经网络组件库（在 PyTorch 生态中，很多这类组件都叫做“模块”）。

从最基础的层面看，神经网络就是若干 *层* 叠在一起。每一层都会对输入执行某种操作，并产生一个输出。

这些层如何组合在一起，取决于你正在处理的问题。

机器学习中最活跃的研究领域之一，就是如何把神经网络的各层组合起来（而且最佳答案一直在变化）。

PyTorch 中绝大多数神经网络组件都包含在 [`torch.nn` 包](https://pytorch.org/docs/stable/nn.html) 中（`nn` 是 neural networks 的缩写）。

```python
from torch import nn
```

### 卷积层（用于构建卷积神经网络，CNN）

PyTorch 提供了 [若干内置卷积层](https://pytorch.org/docs/stable/nn.html#convolution-layers)。

卷积层的命名通常遵循 `torch.nn.ConvXd`，其中 `X` 可以是 `1`、`2` 或 `3`。

`X` 表示卷积操作所作用的维度数量。例如，`1` 表示一维文本，`2` 表示二维图像（高 x 宽），`3` 表示 3D 对象，例如视频（视频可以看作带有时间维度的一系列图像，即 高 x 宽 x 时间）。

> **注意：** 如果你想进一步了解如何使用 PyTorch 构建用于计算机视觉的卷积神经网络，可以查看 [03. PyTorch Computer Vision 第 7.2 节：构建卷积神经网络（CNN）](https://www.learnpytorch.io/03_pytorch_computer_vision/#7-model-2-building-a-convolutional-neural-network-cnn)。

```python
# 创建一个 Conv2d 层（通常用于具有高 x 宽 维度的图像）
conv2d = nn.Conv2d(in_channels=3, # 彩色图像有 3 个通道（红、绿、蓝）
                   out_channels=10,
                   kernel_size=3)
```

### 损失函数

损失函数用于衡量你的模型有多“错”，也就是预测值与真实目标之间偏差有多大。

理想情况下，随着训练数据和优化算法不断作用，损失值应该尽可能下降。

在 PyTorch（以及深度学习）中，损失函数也常被称为：criterion（准则）或 cost function（代价函数）。

PyTorch 在 `torch.nn` 中内置了[多种损失函数](https://pytorch.org/docs/stable/nn.html#loss-functions)。

其中最常见的包括：
* [`nn.L1Loss`](https://pytorch.org/docs/stable/generated/torch.nn.L1Loss.html#torch.nn.L1Loss) - 也称 MAE 或[平均绝对误差](https://en.wikipedia.org/wiki/Mean_absolute_error)（常用于回归问题，或预测房价这类连续数值）。
* [`nn.MSELoss`](https://pytorch.org/docs/stable/generated/torch.nn.MSELoss.html#torch.nn.MSELoss) - 也称 L2Loss 或[均方误差](https://en.wikipedia.org/wiki/Mean_squared_error)（同样常用于回归问题或连续数值预测）。
* [`nn.BCEWithLogitsLoss`](https://pytorch.org/docs/stable/generated/torch.nn.BCEWithLogitsLoss.html#torch.nn.BCEWithLogitsLoss) - 也叫[二元交叉熵](https://en.wikipedia.org/wiki/Cross_entropy)，常用于二分类问题（把样本分成两类之一）。
* [`nn.CrossEntropyLoss`](https://pytorch.org/docs/stable/generated/torch.nn.CrossEntropyLoss.html#torch.nn.CrossEntropyLoss) - 常用于多分类问题（在多个类别中进行分类）。

```python
# L1Loss 损失函数
loss_fn = nn.L1Loss() # also known as MAE or mean absolute error

# MSELoss 损失函数
loss_fn = nn.MSELoss() # also known as MSE or mean squared error

# 二元交叉熵（用于二分类问题）
loss_fn = nn.BCEWithLogitsLoss()

# 交叉熵（用于多分类问题）
loss_fn = nn.CrossEntropyLoss()
```

### 优化器

优化器的任务是调整神经网络权重，从而让损失函数的值不断减小。

PyTorch 在 `torch.optim` 模块中内置了[多种优化算法](https://pytorch.org/docs/stable/optim.html)。

其中两种最常用的优化器包括：
* [`torch.optim.SGD(lr=0.1, params=model.parameters())`](https://pytorch.org/docs/stable/generated/torch.optim.SGD.html#torch.optim.SGD) - SGD 即[随机梯度下降](https://en.wikipedia.org/wiki/Stochastic_gradient_descent)（`lr` 表示学习率，决定每一步更新权重的幅度，值小则调整更细，值大则调整更快）。
* [`torch.optim.Adam(lr=0.001, params=model.parameters())`](https://pytorch.org/docs/stable/generated/torch.optim.Adam.html#torch.optim.Adam) - Adam 优化器（`params` 表示模型参数，也就是训练过程中需要被优化的权重参数）。

## 端到端示例工作流

下面把前面的内容快速串起来，看一个端到端流程示例。

<img src="https://raw.githubusercontent.com/mrdbourke/pytorch-deep-learning/main/images/01_a_pytorch_workflow.png" width=950 alt="从数据到建模、训练再到评估的 PyTorch 工作流"/> 

该流程来源于 [01. PyTorch Workflow Fundamentals](https://www.learnpytorch.io/01_pytorch_workflow/)。

### Create data

```python
# 创建“已知”参数
weight = 0.7
bias = 0.3

# 创建数据
start = 0
end = 1
step = 0.02
X = torch.arange(start, end, step).unsqueeze(dim=1) # data
y = weight * X + bias # labels (want model to learn from data to predict these)

X[:10], y[:10]
```

```python
# 创建训练/测试划分
train_split = int(0.8 * len(X)) # 80% of data used for training set, 20% for testing 
X_train, y_train = X[:train_split], y[:train_split]
X_test, y_test = X[train_split:], y[train_split:]

len(X_train), len(y_train), len(X_test), len(y_test)
```

### 创建模型

在 PyTorch 中创建模型有两种主要方式：
1. 继承 `torch.nn.Module` - 代码更多，但灵活性更强；所有继承 `torch.nn.Module` 的模型都必须实现 `forward()` 方法。
2. 使用 `torch.nn.Sequential` - 代码更少，但灵活性相对较低。

```python
from torch import nn

# 方式 1：继承 torch.nn.Module
class LinearRegressionModel(nn.Module):
    def __init__(self):
        super().__init__()
        # 使用 nn.Linear() 创建模型参数
        self.linear_layer = nn.Linear(in_features=1, 
                                      out_features=1)
    
    # 定义前向计算（输入数据 x 流经 nn.Linear()）
    def forward(self, x: torch.Tensor) -> torch.Tensor:
        return self.linear_layer(x)

model_0 = LinearRegressionModel()
model_0, model_0.state_dict()
```

下面我们使用 `torch.nn.Sequential` 创建一个与上面相同的模型。

```python
from torch import nn

# 方式 2：使用 torch.nn.Sequential
model_1 = torch.nn.Sequential(
    nn.Linear(in_features=1,
              out_features=1))

model_1, model_1.state_dict()
```

### 配置损失函数与优化器

```python
# 创建损失函数
loss_fn = nn.L1Loss()

# 创建优化器
optimizer = torch.optim.SGD(params=model_1.parameters(), # optimize newly created model's parameters
                            lr=0.01)
```

### 创建训练/测试循环

我们的目标是降低模型损失（也就是模型预测与真实数据之间的差距）。

如果训练/测试循环实现正确，且模型确实能从数据中学习到模式，那么训练损失和测试损失都应该逐步下降。

关于 PyTorch 训练循环，可参考以下资源：
* [PyTorch 优化循环歌曲](https://youtu.be/Nutpusq_AFw)
* [PyTorch 工作流基础 第 3 节：训练循环](https://www.learnpytorch.io/01_pytorch_workflow/#pytorch-training-loop)
* [PyTorch 工作流基础 第 3 节：测试循环](https://www.learnpytorch.io/01_pytorch_workflow/#pytorch-testing-loop)
* [PyTorch 工作流基础 第 6.3 节：训练](https://www.learnpytorch.io/01_pytorch_workflow/#63-training)

```python
torch.manual_seed(42)

# 设置训练轮数
epochs = 1000 

# 将数据放到可用设备上
# 若不这样做会报错（并非所有数据都在目标设备上）
X_train = X_train.to(device)
X_test = X_test.to(device)
y_train = y_train.to(device)
y_test = y_test.to(device)

# 将模型放到可用设备上
# 若不这样做会报错（模型不在目标设备上）
model_1 = model_1.to(device)

for epoch in range(epochs):
    ### Training
    model_1.train() # train mode is on by default after construction

    # 1. Forward pass
    y_pred = model_1(X_train)

    # 2. Calculate loss
    loss = loss_fn(y_pred, y_train)

    # 3. Zero grad optimizer
    optimizer.zero_grad()

    # 4. Loss backward
    loss.backward()

    # 5. Step the optimizer
    optimizer.step()

    ### Testing
    model_1.eval() # put the model in evaluation mode for testing (inference)
    # 1. Forward pass
    with torch.inference_mode():
        test_pred = model_1(X_test)
    
        # 2. Calculate the loss
        test_loss = loss_fn(test_pred, y_test)

    if epoch % 100 == 0:
        print(f"Epoch: {epoch} | Train loss: {loss} | Test loss: {test_loss}")
```
